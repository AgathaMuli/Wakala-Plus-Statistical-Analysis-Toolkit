## Question 7: What report should be saved for this project?


```python
from pathlib import Path
from stats_toolkit import StatisticsAnalyzer, DATA_PATH

transactions = StatisticsAnalyzer.from_csv(DATA_PATH, "transactions")
deposits = StatisticsAnalyzer.from_csv(DATA_PATH, "daily_deposits_kes")
report = "\n".join([
    "Statistical Summary Report",
    "========================",
    "",
    "Transactions (with outliers):",
    transactions.report(),
    "",
    "Transactions (without outliers):",
    transactions.without_outliers().report(),
    "",
    "Daily Deposits (with outliers):",
    deposits.report(),
    "",
    "Daily Deposits (without outliers):",
    deposits.without_outliers().report(),
])
output_path = Path("output") / "stats_report.txt"
output_path.parent.mkdir(exist_ok=True, parents=True)
output_path.write_text(report, encoding="utf-8")
print(report)
print(f"\nReport written to: {output_path}")
```
